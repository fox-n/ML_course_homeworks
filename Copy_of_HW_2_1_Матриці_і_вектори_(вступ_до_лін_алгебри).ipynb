```python
A = np.array(
    [[0.18, 0.53],
     [0.75, 0.47],
     [0.35, 0.21]])

B = np.array(
    [[0.17, 0.58, 0.75],
     [0.38, 0.11, 0.15]])

C = np.array([[0], [1]])

D = np.array([[0,1], [1,0]])

display(A, B, C, D)
```

```python
#1 Множення матриць:

result_matrix = np.dot(B, A)
print(f'Результуюча матриця:\n {result_matrix}')
```

```python
print(f'Результат множення матриці result_matrix на матрицю С:\n{np.dot(result_matrix, C)}')
```

```python
print(f'Результат множення матриці result_matrix на матрицю D:\n{np.dot(result_matrix, D)}')
```
